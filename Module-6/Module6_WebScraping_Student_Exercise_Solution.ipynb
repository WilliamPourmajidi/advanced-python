# Module 6: In-Class Activity - Web Scraping (Solution)
This notebook contains one complete solution for scraping country name, capital, and population from Scrape This Site.

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

## Step 1: Helper function to fetch and parse the page

```python
def get_soup():
    """Fetch the countries page and return a BeautifulSoup object."""
    url = "https://www.scrapethissite.com/pages/simple/"

    response = requests.get(url, timeout=20)
    print("Status Code:", response.status_code)
    if response.status_code != 200:
        return None

    return BeautifulSoup(response.text, 'html.parser')
```

```python
soup = get_soup()

if soup is not None:
    print("Title:", soup.title.text.strip())
    country_blocks = soup.find_all('div', class_='country')
    print("Country blocks found:", len(country_blocks))
```

## Step 2: Parse one country block

```python
def parse_country_block(block):
    country_name = block.find('h3', class_='country-name').get_text(strip=True)
    capital = block.find('span', class_='country-capital').get_text(strip=True)
    population_text = block.find('span', class_='country-population').get_text(strip=True)
    population = int(population_text)

    return {
        'Country': country_name,
        'Capital': capital,
        'Population': population
    }
```

## Step 3: Scrape all country blocks

```python
all_countries = []

if soup is not None:
    country_blocks = soup.find_all('div', class_='country')
    for block in country_blocks:
        all_countries.append(parse_country_block(block))

print("Total countries scraped:", len(all_countries))
```

## Step 4: Convert to DataFrame

```python
df_countries = pd.DataFrame(all_countries)
df_countries.head(10)
```

## Step 5: Mini analysis tasks

```python
# Task 1: Top 10 countries by population
top_10_population = df_countries.sort_values('Population', ascending=False).head(10)
print("Top 10 countries by population:")
print(top_10_population[['Country', 'Population']])

# Task 2: Population of Canada
canada_population = df_countries.loc[df_countries['Country'] == 'Canada', 'Population']
if not canada_population.empty:
    print("\nPopulation of Canada:", int(canada_population.iloc[0]))

# Task 3: Save CSV
output_path = 'countries_activity_output.csv'
df_countries.to_csv(output_path, index=False)
print("\nSaved file:", output_path)
```

## Bonus (Solved)

```python
bonus_df = df_countries.copy()
bonus_df['Population_Millions'] = (bonus_df['Population'] / 1_000_000).round(2)

above_50m = bonus_df[bonus_df['Population'] > 50_000_000]
print("Countries above 50 million population:", len(above_50m))
print(above_50m[['Country', 'Capital', 'Population_Millions']].head(10))

most_populated = bonus_df.sort_values('Population', ascending=False).iloc[0]
print("\nMost populated country:", most_populated['Country'])
print("Capital city:", most_populated['Capital'])
```
