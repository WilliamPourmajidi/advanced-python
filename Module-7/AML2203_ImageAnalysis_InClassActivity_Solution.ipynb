## Learning Goals

This solution demonstrates how to load, inspect, transform, analyze, and save images using `PIL`, `matplotlib`, and `numpy`.

## 1. Import Libraries

```python
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np
```

## 2. Load and Display an Image

```python
image_path = 'main_image.jpg'
image = Image.open(image_path)

plt.figure(figsize=(8, 6))
plt.imshow(image)
plt.axis('off')
plt.title('Original Image')
plt.show()
```

## 3. Inspect Image Information

```python
print('Image Format:', image.format)
print('Image Mode  :', image.mode)
print('Image Size  :', image.size)
```

## 4. Convert the Image to a NumPy Array

```python
image_array = np.array(image)

print('Shape of image array:', image_array.shape)
print('Data type:', image_array.dtype)

image_array[:5, :5]
```

## 5. Convert to Grayscale

```python
gray_image = image.convert('L')

plt.figure(figsize=(8, 6))
plt.imshow(gray_image, cmap='gray')
plt.axis('off')
plt.title('Grayscale Image')
plt.show()
```

## 6. Crop the Image

```python
crop_box = (210, 320, 480, 450)
cropped = image.crop(crop_box)

plt.figure(figsize=(8, 4))
plt.imshow(cropped)
plt.axis('off')
plt.title('Cropped Image')
plt.show()
```

## 7. Split RGB Channels

```python
r, g, b = image.split()
channels = [(r, 'Red Channel'), (g, 'Green Channel'), (b, 'Blue Channel')]

plt.figure(figsize=(12, 4))
for index, (channel, title) in enumerate(channels, start=1):
    plt.subplot(1, 3, index)
    plt.imshow(channel, cmap='gray')
    plt.axis('off')
    plt.title(title)
plt.show()
```

## 8. Grayscale Pixel Intensity Histogram

```python
gray_array = np.array(gray_image)
gray_flat = gray_array.flatten()

plt.figure(figsize=(8, 5))
plt.hist(gray_flat, bins=256, range=(0, 255), color='gray')
plt.title('Pixel Intensity Histogram')
plt.xlabel('Pixel Intensity')
plt.ylabel('Frequency')
plt.grid(True)
plt.show()
```

## 9. Basic Pixel Statistics

```python
rgb_mean = np.mean(image_array, axis=(0, 1))
rgb_std = np.std(image_array, axis=(0, 1))

print('Mean pixel value (R, G, B):', rgb_mean)
print('Standard deviation (R, G, B):', rgb_std)
```

## 10. Highlight the Most Common Grayscale Intensity

```python
counts = np.bincount(gray_flat)
most_frequent_intensity = np.argmax(counts)
print('Most frequent pixel intensity:', most_frequent_intensity)

mask = gray_array == most_frequent_intensity
highlighted = np.stack([gray_array] * 3, axis=-1)
highlighted[mask] = [255, 0, 0]

plt.figure(figsize=(8, 6))
plt.imshow(highlighted)
plt.axis('off')
plt.title(f'Pixels with Intensity = {most_frequent_intensity} Highlighted in Red')
plt.show()
```

## 11. Resize, Rotate, and Save

```python
width, height = image.size
new_size = (width // 2, height // 2)
resized = image.resize(new_size)

rotated = resized.rotate(90, expand=True)
rotated.save('student_processed_image.jpg')

plt.figure(figsize=(6, 6))
plt.imshow(rotated)
plt.axis('off')
plt.title('Resized and Rotated Image')
plt.show()
```
